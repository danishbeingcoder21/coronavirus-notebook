```python
#Q12=>>>five countries with highest number of covid cases
df.nlargest(5,'Total Cases')
#nlargest() sbse badi values nikalata hai
```

```python
#Q13=>>> bar plot of top 5 countries with Highest cases
top5 = df.nlargest(5,'Total Cases')
plt.figure(figsize=(8,5))
sns.barplot(x='Country',y='Total Cases', data=top5)
plt.title('Top  Countries with Highest COVID Cases')
plt.xticks(rotation=45)
plt.show()
#plt.figure(figsize=(8,5)): graph ka size set krta hai
#sns.barplot(): graph banata hai
#x='Country': countries x axis pr
#y='Total Cases': cases y-axis pr
#plt.show(): graph display krta hai 
```

```python
#q14=>>> Five countries with lowest COVID Cases
df.nsmallest(5,'Total Cases')
#nsmallest(): sbse chotii value return krta hai
```

```python
#q15=>> Barplot of countries with lowest Cases
bottom5 = df.nsmallest(5,'Total Cases')
plt.figure(figsize=(8,5))
sns.barplot(x='Country',y='Total Cases',data=bottom5)
plt.title('5 Countries with Lowet COVID Cases')
plt.xticks(rotation=45)
plt.show()
```

```python
#Q16=>> find the five countris hat have the highest number of Covid-19 deaths
top5_deaths =df.nlargest(5,'Total Deaths')
top5_deaths
#nlargest(): largest values nikalata hai 
```

```python
#Q17=>>>>crate a bar plot showing the top five countries with the highest toatl COVID-19 cases
top5_deaths = df.nlargest(5,'Total Deaths')
plt.figure(figsize=(10,5))
sns.barplot(x='Country',y='Total Deaths',data=top5_deaths)
plt.title('Top 5 countries with Highest COVID-19 Deaths')
plt.xticks(rotation=45)
plt.show()
#sns.barplot() bar graph banata hai.
#Country x-axis par show hoga.
#Total Deaths y-axis par show hongi.
#Graph se comparison easy ho jaata hai.
```

```python
#q19=>> Find the five countries with the highest Covid-19 death percentage
top5_death_percentage = df.nlargest(5,'Death percentage')
top5_death_percentage
#death percentage formula 
(Total Deaths / Total Cases) × 100
```

```python
#Q20=>>>Create a bar plot showing the top five countries with highest Covid-19 death percentage
top5_death_percentage=df.nlargest(5,'Death percentage')
plt.figure(figsize=(10,5))
sns.barplot(x='Country',y='Death percentage',data=top5_death_percentage)
plt.title('Top 5 Countries with Highest COVID-19 Death Percentage')
plt.xticks(rotation=45)
plt.show()
#X-axis → Country names
#Y-axis → Death percentage
#Bar ki height death percentage ko represent karegi.
#Ye graph batata hai ki kin countries mein death rate sabse zyada tha.
```

```python
#q21=>>> find the five countries with the lowest COVID-19 death percentage
lowest_death_percentage =df.nsmallest(5,'Death percentage')
lowest_death_percentage
```

```python
#q22=>>Create a bar plot for five countries with the lowest death percentage
lowest_death_percentage=df.nsmallest(5,'Death percentage')
plt.figure(figsize=(10,5))
sns.barplot(x='Country',y='Death percentage',data=lowest_death_percentage)
plt.title('Top 5 Countries with Lowest Death Percentage')
plt.xticks(rotation=45)
plt.show()
#X-axis → Country
#Y-axis → Death Percentage
#Lowest death rate wale countries graph mein show honge.
```

```python
#q23=>>> Create a scatter PLot between Total cases and Total Deaths
plt.figure(figsize=(10,6))
sns.scatterplot(x='Total Cases',y= 'Total Deaths',data=df)
plt.title('Total Cases vs Total Deaths')
plt.show()
#Scatter plot do numerical columns ke beech relationship dikhata hai.
```

```python
#q24=>>> dISPLAY the correlation Matrix
correlation_matrix=df.corr(numeric_only=True)
correlation_matrix
#corr():numeric columns ke beech relationship measure karta hai
```

```python
#q25=>> Create a heastmap of correlation Matrix
plt.figure(figsize=(10,8))
correlation_matrix = df.corr(numeric_only=True)
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
#snsheatmap():correlation values ko colors ki form mein show karta hai.
#annot=True: heat maps ke andr numbers bhi dikhata hai
#cmap='coolwarm':color theme set krta hai
```
